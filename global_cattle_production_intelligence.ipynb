# Data View & import

```python
# 1. Import Library Utama
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Atur tampilan visualisasi agar rapi
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# 2. Muat Dataset
print("⏳ Sedang memuat dataset...")
df = pd.read_csv("global_cattle_disease_detection_dataset.csv")

print(f"✅ Berhasil dimuat! Dimensi Data: {df.shape[0]:,} Baris x {df.shape[1]} Kolom")

# 3. Pratinjau 5 Baris Pertama
display(df.head())
```

# Step 2 Dasbor Preprocessing & EDA (Univariate Analysis)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings

# Suppress warnings
warnings.filterwarnings('ignore')

# Set style untuk visualisasi yang lebih menarik
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("="*60)
print("🐄 ANALISIS DATA PETERNAKAN SAPI PERAH 🥛")
print("="*60)

# 1. Ubah kolom Date menjadi tipe datetime
df["Date"] = pd.to_datetime(df["Date"])
print("✅ Konversi kolom Date ke datetime berhasil!")

# 2. Periksa Missing Values dengan tampilan yang lebih detail
missing_data = df.isnull().sum()
total_missing = missing_data.sum()

print(f"\n🔍 ANALISIS MISSING VALUES")
print("-" * 40)
if total_missing == 0:
    print("🎉 Excellent! Tidak ada missing values dalam dataset")
else:
    print(f"⚠️  Total Missing Values: {total_missing}")
    for col, missing in missing_data[missing_data > 0].items():
        print(f"   • {col}: {missing} values")

# 3. Feature Engineering - Feed Efficiency
df["Feed_Efficiency"] = df["Milk_Yield_L"] / df["Feed_Quantity_kg"]
print(f"\n🧮 FEATURE ENGINEERING")
print("-" * 40)
print("✅ Fitur 'Feed_Efficiency' berhasil dibuat!")
print(f"📈 Rata-rata efisiensi pakan: {df['Feed_Efficiency'].mean():.2f} L/kg")

# 4. Health Status Analysis
df["Is_Sick"] = df["Disease_Status"].apply(lambda x: 0 if x == "Healthy" else 1)

print(f"\n🏥 ANALISIS STATUS KESEHATAN SAPI")
print("-" * 40)

# Hitung proporsi dengan tampilan yang lebih menarik
health_stats = df["Is_Sick"].value_counts()
health_props = df["Is_Sick"].value_counts(normalize=True) * 100

print(f"🟢 Sapi Sehat (0):  {health_stats[0]:,} ekor ({health_props[0]:.1f}%)")
print(f"🔴 Sapi Sakit (1):  {health_stats[1]:,} ekor ({health_props[1]:.1f}%)")

# Visualisasi tanpa emoji di plot (menghindari warning font)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Pie chart untuk status kesehatan - tanpa emoji
labels = ['Sehat', 'Sakit']
colors = ['#2ecc71', '#e74c3c']
wedges, texts, autotexts = ax1.pie(health_stats.values, labels=labels, autopct='%1.1f%%', 
                                   colors=colors, startangle=90)

# Styling untuk pie chart
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(10)

ax1.set_title('Distribusi Status Kesehatan Sapi', fontsize=14, fontweight='bold', pad=20)

# Histogram untuk Feed Efficiency
n, bins, patches = ax2.hist(df['Feed_Efficiency'], bins=20, color='skyblue', 
                           alpha=0.7, edgecolor='black', linewidth=0.8)

# Color gradient untuk histogram
for i, patch in enumerate(patches):
    patch.set_facecolor(plt.cm.viridis(i / len(patches)))

ax2.set_title('Distribusi Efisiensi Pakan', fontsize=14, fontweight='bold', pad=20)
ax2.set_xlabel('Feed Efficiency (L/kg)', fontsize=12)
ax2.set_ylabel('Frekuensi', fontsize=12)
ax2.grid(True, alpha=0.3)

# Tambahkan statistik pada histogram
mean_eff = df['Feed_Efficiency'].mean()
ax2.axvline(mean_eff, color='red', linestyle='--', linewidth=2, 
           label=f'Rata-rata: {mean_eff:.2f}')
ax2.legend()

plt.tight_layout()
plt.show()

print(f"\n📋 RINGKASAN DATASET")
print("-" * 40)
print(f"📅 Periode data: {df['Date'].min().strftime('%d/%m/%Y')} - {df['Date'].max().strftime('%d/%m/%Y')}")
print(f"🐄 Total sapi: {len(df):,} records")
print(f"📊 Jumlah kolom: {len(df.columns)}")
print("="*60)
```
